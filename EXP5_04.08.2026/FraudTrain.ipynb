```
import os
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Target file path
file_path = '/content/fraudTrain.csv'

if not os.path.exists(file_path):
    print(f"Error: File not found at {file_path}")
else:
    print(f"\n{'='*20} Processing: {os.path.basename(file_path)} {'='*20}")

    # 1. Load the Dataset
    df_train = pd.read_csv(file_path)

    # 2. Binning (Discretisation)
    amt_bins = [0, 50, 500, 2000, np.inf]
    bin_labels = ['Low', 'Medium', 'High', 'Very High']
    df_train['Amt_Category'] = pd.cut(df_train['amt'], bins=amt_bins, labels=bin_labels)

    print("\n Binning Distribution Counts:")
    print(df_train['Amt_Category'].value_counts())

    # 3. Outlier Detection and Filtering
    outlier_threshold = 3000
    outliers = df_train[df_train['amt'] > outlier_threshold]
    print(f"\n Outlier Detection:")
    print(f"Total outliers exceeding ${outlier_threshold}: {len(outliers)}")
    if len(outliers) > 0:
        print("Sample outlier rows:")
        print(outliers[['cc_num', 'amt', 'merchant']].head(3))

    # 4. Data Transformation (Feature Scaling)
    min_max_scaler = MinMaxScaler()
    df_train['amt_normalized'] = min_max_scaler.fit_transform(df_train[['amt']])

    standard_scaler = StandardScaler()
    df_train['amt_standardized'] = standard_scaler.fit_transform(df_train[['amt']])

    print("\n Data Transformation (Feature Scaling Samples):")
    print(df_train[['amt', 'amt_normalized', 'amt_standardized']].head(5))
```
