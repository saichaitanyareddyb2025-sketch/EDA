```python
import pandas as pd
import numpy as np

# ==========================================
# 0. DATASET RECONSTRUCTION (From Image)
# ==========================================
# Extracting a subset of rows from the image to serve as our main listings dataset
listings_data = {
    'listing_id': [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    'name': [
        "Beautiful Flat in le Village Montmartre, Paris",
        "39 m² Paris (Sacre Cœur)",
        "Lovely apartment with Terrace, 60m2",
        "Cosy studio (close to Eiffel tower)",
        "Close to Eiffel Tower - Beautiful flat : 2 rooms",
        "NEW - Charming Apartment le Marais",
        "2p - Entre Bastille et Republique"
    ],
    'host_id': [2001, 2002, 2003, 2004, 2005, 2006, 2007],
    'host_since': ["2011-12-03", "2013-11-29", "2014-07-31", "2013-12-17", "2014-12-14", "2011-04-13", "2013-08-09"],
    'host_location': ["Ile-de-France, France", "Ile-de-France, France", "Ile-de-France, France", "Ile-de-France, France", "Ile-de-France, France", "Ile-de-France, France", "Ile-de-France, France"]
}
df_listings = pd.DataFrame(listings_data)

# Creating a secondary mock dataset representing Review metrics to merge with our image dataset
reviews_data = {
    'listing_id': [1001, 1003, 1008, 1004, 9999999], # 9999999 represents a listing not in df_listings
    'review_scores_rating': [95, 88, 70, 92, 65],
    'number_of_reviews': [24, 118, 5, 42, 1]
}
df_reviews = pd.DataFrame(reviews_data)


# ==========================================
# DATASET MERGING DEMONSTRATION
# ==========================================
print("--- Image Dataset (Listings Sub-sample) ---")
print(df_listings)

# Inner Join: Only includes listings present in both dataframes
listings_inner = pd.merge(df_listings, df_reviews, on='listing_id', how='inner')
print("\nMerged Dataset (Inner Join on listing_id):")
print(listings_inner)

# Left Join: Keeps all listings from the image dataset, filling missing review data with NaN
listings_left = pd.merge(df_listings, df_reviews, on='listing_id', how='left')
print("\nMerged Dataset (Left Join on listing_id):")
print(listings_left)


# ==========================================
# 1. CONCATENATION OF DATAFRAMES
# ==========================================
dataFrame1 = pd.DataFrame({
    'StudentID': [1, 2, 3, 4, 5],
    'Score': [89, 39, 50, 97, 22]
})

dataFrame2 = pd.DataFrame({
    'StudentID': [6, 7, 8, 9, 10],
    'Score': [98, 93, 44, 77, 69]
})

# Vertical Concatenation
vertical_concat = pd.concat([dataFrame1, dataFrame2], ignore_index=True)
print("\nVertical Concatenation:")
print(vertical_concat)

# Horizontal Concatenation
horizontal_concat = pd.concat([dataFrame1, dataFrame2], axis=1)
print("\nHorizontal Concatenation:")
print(horizontal_concat)


# ==========================================
# 2. MERGING DATAFRAMES
# ==========================================
dfSE = pd.DataFrame({
    'StudentID': [1, 2, 3, 4, 5, 6],
    'ScoreSE': [98, 93, 44, 77, 69, 56]
})

dfML = pd.DataFrame({
    'StudentID': [3, 4, 7, 8, 9, 10, 11, 12],
    'ScoreML': [39, 93, 49, 44, 55, 78, 97, 87]
})

# Inner Join
inner_join = pd.merge(dfSE, dfML, on='StudentID', how='inner')
print("\nInner Join:")
print(inner_join)

# Left Join
left_join = pd.merge(dfSE, dfML, on='StudentID', how='left')
print("\nLeft Join:")
print(left_join)

# Right Join
right_join = pd.merge(dfSE, dfML, on='StudentID', how='right')
print("\nRight Join:")
print(right_join)

# Outer Join
outer_join = pd.merge(dfSE, dfML, on='StudentID', how='outer')
print("\nOuter Join:")
print(outer_join)


# ==========================================
# 3. ADDING A NEW COLUMN
# ==========================================
sales = pd.DataFrame({
    'Quantity': [10, 15, 20, 25],
    'UnitPrice': [50, 60, 45, 80]
})

sales['TotalPrice'] = sales['Quantity'] * sales['UnitPrice']
print("\nSales Data:")
print(sales)


# ==========================================
# 4. DESCRIPTIVE STATISTICS
# ==========================================
print("\nDescriptive Statistics:")
print(sales.describe())


# ==========================================
# 5. VALUE COUNTS
# ==========================================
company_df = pd.DataFrame({
    'Company': ['ABC', 'XYZ', 'ABC', 'PQR', 'XYZ', 'ABC']
})
print("\nCompany Counts:")
print(company_df['Company'].value_counts())


# ==========================================
# 6. HIERARCHICAL INDEXING (STACK/UNSTACK)
# ==========================================
data = np.arange(15).reshape((3, 5))
dframe = pd.DataFrame(
    data,
    index=['Rainfall', 'Humidity', 'Wind'],
    columns=['Bergen', 'Oslo', 'Trondheim', 'Stavanger', 'Kristiansand']
)
print("\nOriginal DataFrame:")
print(dframe)

# Stack
stacked = dframe.stack()
print("\nStacked Data:")
print(stacked)

# Unstack
unstacked = stacked.unstack()
print("\nUnstacked Data:")
print(unstacked)


# ==========================================
# 7. SERIES CONCATENATION WITH KEYS
# ==========================================
series1 = pd.Series([0, 111, 222, 333], index=['zeros', 'ones', 'twos', 'threes'])
series2 = pd.Series([444, 555, 666], index=['fours', 'fives', 'sixs'])

frame = pd.concat([series1, series2], keys=['Number1', 'Number2'])
print("\nHierarchical Series:")
print(frame)

print("\nUnstacked Series:")
print(frame.unstack())
```
