## Handling Missing Data - Filling Strategies

```
import pandas as pd
import numpy as np
from IPython.display import display # Import display function

# Recreate the initial DataFrame with missing values as described in the context
data = np.arange(15, 30).reshape(5, 3)
dfx = pd.DataFrame(data, index=['apple', 'banana', 'kiwi',
'grapes', 'mango'], columns=['store1', 'store2', 'store3'])

# Add missing values
dfx['store4'] = np.nan
dfx.loc['watermelon'] = np.arange(15, 19)
dfx.loc['oranges'] = np.nan
dfx['store5'] = np.nan
dfx.loc['apple', 'store4'] = 20.0 # Fix for ChainedAssignmentError warning

print("Original DataFrame with NaNs:")
display(dfx)
```

### Strategy 1: Filling with a constant value

```
# Fill all NaN values with a constant (e.g., 0) using .replace() for a different approach
# Note: While df.fillna(0) is common, df.replace(np.nan, 0) also achieves the same for NaNs.
filled_df_replace = dfx.replace(np.nan, 0)
print("DataFrame after filling NaNs with 0 using .replace():")
display(filled_df_replace)

# Compare means to the original dfx to see the impact of filling
print("Mean of original dfx:\n", dfx.mean())
print("Mean of filled_df_replace:\n", filled_df_replace.mean())
```

### Strategy 2: Forward and Backward Filling

```
# Forward-fill (ffill) the 'store4' column
ffilled_store4 = dfx['store4'].ffill()
print("Forward-filled 'store4' column:")
display(ffilled_store4)

# Backward-fill (bfill) the 'store4' column
bf_filled_store4 = dfx['store4'].bfill()
print("Backward-filled 'store4' column:")
display(bf_filled_store4)
```

### Strategy 3: Filling with statistical measures (e.g., mean, median)

```
# Fill missing values in 'store5' with its mean
# Note: This is an alternative to filling with a constant 0, often used for numerical data.

# Calculate the mean of 'store5' excluding NaNs
mean_store5 = dfx['store5'].mean()

# Fill NaNs in 'store5' with its mean
dfx_filled_mean = dfx.copy() # Work on a copy to avoid modifying original dfx
dfx_filled_mean['store5'] = dfx_filled_mean['store5'].fillna(mean_store5)

print(f"DataFrame after filling 'store5' NaNs with its mean ({mean_store5:.2f}):")
display(dfx_filled_mean)

# Fill all numerical columns with their respective means
dfx_filled_all_means = dfx.fillna(dfx.mean(numeric_only=True))
print("DataFrame after filling all numerical NaNs with column means:")
display(dfx_filled_all_means)
```
